```
class Product
{
    public string Name;
    public double Price;
    public string Manufacturer;

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}";
    }

    public virtual void Discount(double percent)
    {
        Price = Price - (Price * percent / 100);
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }
}

class Electronics : Product
{
    public int WarrantyPeriod;

    public override void Discount(double percent)
    {
        if (WarrantyPeriod > 24)
        {
            percent = percent + 5;
        }
        Price = Price - (Price * percent / 100);
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес.";
    }
}

class Clothing : Product
{
    public string Size;

    public override void Display()
    {
        Console.WriteLine(GetInfo() + $", Размер: {Size}");
    }
}

class Books : Product
{
    public string Author;

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}";
    }
}

Product myProduct = new Product();
myProduct.Name = "Стол";
myProduct.Price = 5000;
myProduct.Manufacturer = "МебельПром";
myProduct.Display();

Electronics myElectronics = new Electronics();
myElectronics.Name = "Телевизор";
myElectronics.Price = 50000;
myElectronics.Manufacturer = "Samsung";
myElectronics.WarrantyPeriod = 36;
myElectronics.Discount(10);
myElectronics.Display();

Clothing myClothing = new Clothing();
myClothing.Name = "Футболка";
myClothing.Price = 1500;
myClothing.Manufacturer = "Nike";
myClothing.Size = "L";
myClothing.Display();

Books myBook = new Books();
myBook.Name = "Война и мир";
myBook.Price = 1200;
myBook.Manufacturer = "Эксмо";
myBook.Author = "Лев Толстой";
myBook.Display();
```
