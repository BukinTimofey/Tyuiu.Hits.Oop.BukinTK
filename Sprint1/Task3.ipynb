```
class ProductV2
{
    private string name;
    private double price;
    private string manufacturer;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public double Price
    {
        get { return price; }
        set { price = value; }
    }

    public string Manufacturer
    {
        get { return manufacturer; }
        set { manufacturer = value; }
    }

    public ProductV2(string name, double price, string manufacturer)
    {
        this.name = name;
        this.price = price;
        this.manufacturer = manufacturer;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}";
    }

    public virtual void Discount(double percent)
    {
        Price = Price - (Price * percent / 100);
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }
}

class ElectronicsV2 : ProductV2
{
    private int warrantyPeriod;

    public int WarrantyPeriod
    {
        get { return warrantyPeriod; }
        set { warrantyPeriod = value; }
    }

    public ElectronicsV2(string name, double price, string manufacturer, int warrantyPeriod)
        : base(name, price, manufacturer)
    {
        this.warrantyPeriod = warrantyPeriod;
    }

    public override void Discount(double percent)
    {
        if (WarrantyPeriod > 24)
        {
            percent = percent + 5;
        }
        Price = Price - (Price * percent / 100);
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес.";
    }
}

class ClothingV2 : ProductV2
{
    private string size;

    public string Size
    {
        get { return size; }
        set { size = value; }
    }

    public ClothingV2(string name, double price, string manufacturer, string size)
        : base(name, price, manufacturer)
    {
        this.size = size;
    }

    public override void Display()
    {
        Console.WriteLine(GetInfo() + $", Размер: {Size}");
    }
}

class BooksV2 : ProductV2
{
    private string author;

    public string Author
    {
        get { return author; }
        set { author = value; }
    }

    public BooksV2(string name, double price, string manufacturer, string author)
        : base(name, price, manufacturer)
    {
        this.author = author;
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}";
    }
}

// Очень много очень кода

class StoreV2
{
    private string storeName;
    private List<ProductV2> products;

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public List<ProductV2> Products
    {
        get { return products; }
        set { products = value; }
    }

    public StoreV2(string storeName)
    {
        this.storeName = storeName;
        products = new List<ProductV2>();
    }

    public void AddProduct(ProductV2 product)
    {
        products.Add(product);
    }

    public void ApplyDiscountToAll(double percent)
    {
        foreach (ProductV2 product in products)
        {
            product.Discount(percent);
        }
    }

    public void ShowAllProducts()
    {
        Console.WriteLine($"Магазин: {StoreName}");
        foreach (ProductV2 product in products)
        {
            product.Display();
        }
    }
}

ElectronicsV2 myElectronicsV2 = new ElectronicsV2("Телик", 50000, "SaSun", 36);
ClothingV2 myClothingV2 = new ClothingV2("Футфолка", 1500, "Nike", "L");
BooksV2 myBookV2 = new BooksV2("Война и мир", 1200, "ООО рога и копыта", "Лев Толстой");
StoreV2 myStoreV2 = new StoreV2("Товары V2");

myStoreV2.AddProduct(myElectronicsV2);
myStoreV2.AddProduct(myClothingV2);
myStoreV2.AddProduct(myBookV2);
myStoreV2.ShowAllProducts();

Console.WriteLine("--- Применяем скидку 10(1)% ---");
myStoreV2.ApplyDiscountToAll(10);
myStoreV2.ShowAllProducts();
// То что написано выше не оцень похоже на ООП на самом деле
```
