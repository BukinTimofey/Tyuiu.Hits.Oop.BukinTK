```
public interface IShippable
{
    void Ship(string address);
    double CalculateShippingCost();
}

public interface IReviewable
{
    void AddReview(string review);
    List<string> GetReviews();
}

public interface IWarrantyService
{
    void RequestWarrantyRepair();
    string GetWarrantyStatus();
}

class ProductV3
{
    private string name;
    private double price;
    private string manufacturer;
    private string id;
    private string category;
    private int stockQuantity;
    private double rating;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }

    public double Price
    {
        get { return price; }
        set { price = value; }
    }

    public string Manufacturer
    {
        get { return manufacturer; }
        set { manufacturer = value; }
    }

    public string Id
    {
        get { return id; }
        set { id = value; }
    }

    public string Category
    {
        get { return category; }
        set { category = value; }
    }

    public int StockQuantity
    {
        get { return stockQuantity; }
        set { stockQuantity = value < 0 ? 0 : value; }
    }

    public double Rating
    {
        get { return rating; }
        set { rating = (value >= 0 && value <= 5) ? value : 0; }
    }

    public ProductV3(string name, double price, string manufacturer, string id, string category, int stockQuantity)
    {
        this.name = name;
        this.price = price;
        this.manufacturer = manufacturer;
        this.id = id;
        this.category = category;
        this.stockQuantity = stockQuantity;
        this.rating = 0;
    }

    public virtual string GetInfo()
    {
        return $"Название: {Name}, Цена: {Price} руб., Производитель: {Manufacturer}, ID: {Id}, Категория: {Category}";
    }

    public virtual void Discount(double percent)
    {
        Price = Price - (Price * percent / 100);
    }

    public virtual void Display()
    {
        Console.WriteLine(GetInfo());
    }

    public void Restock(int amount)
    {
        StockQuantity = StockQuantity + amount;
        Console.WriteLine($"{Name}: склад пополнен на {amount} шт. Теперь на складе: {StockQuantity}.");
    }

    public bool IsAvailable()
    {
        return StockQuantity > 0;
    }

    public void UpdateRating(double newRating)
    {
        Rating = newRating;
        Console.WriteLine($"{Name}: оценка обновлена — {Rating} из 5.");
    }

    public string GetStockStatus()
    {
        if (StockQuantity == 0) return "Нет в наличии";
        if (StockQuantity < 5) return "Мало на складе!!!";
        return "В наличии";
    }
}

class ElectronicsV3 : ProductV3
{
    private int warrantyPeriod;
    private int powerConsumption;
    private string countryOfOrigin;
    private bool isSmart;

    public int WarrantyPeriod
    {
        get { return warrantyPeriod; }
        set { warrantyPeriod = value; }
    }

    public int PowerConsumption
    {
        get { return powerConsumption; }
        set { powerConsumption = value < 0 ? 0 : value; }
    }

    public string CountryOfOrigin
    {
        get { return countryOfOrigin; }
        set { countryOfOrigin = value; }
    }

    public bool IsSmart
    {
        get { return isSmart; }
        set { isSmart = value; }
    }

    public ElectronicsV3(string name, double price, string manufacturer, string id, string category, int stockQuantity, int warrantyPeriod, int powerConsumption, string countryOfOrigin, bool isSmart)
        : base(name, price, manufacturer, id, category, stockQuantity)
    {
        this.warrantyPeriod = warrantyPeriod;
        this.powerConsumption = powerConsumption;
        this.countryOfOrigin = countryOfOrigin;
        this.isSmart = isSmart;
    }

    public override void Discount(double percent)
    {
        if (WarrantyPeriod > 24)
        {
            percent = percent + 5;
        }
        Price = Price - (Price * percent / 100);
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Гарантия: {WarrantyPeriod} мес.";
    }

    public string GetPowerInfo()
    {
        return $"{Name}: потребление {PowerConsumption} Вт, страна — {CountryOfOrigin}";
    }

    public bool IsEnergyEfficient()
    {
        return PowerConsumption < 100;
    }

    public void ToggleSmartMode()
    {
        IsSmart = !IsSmart;
        Console.WriteLine($"{Name}: умный режим {(IsSmart ? "включён!!!" : "выключен")}.");
    }
}

class ClothingV3 : ProductV3, IShippable
{
    private string size;
    private string material;
    private string color;
    private string gender;

    public string Size
    {
        get { return size; }
        set { size = value; }
    }

    public string Material
    {
        get { return material; }
        set { material = value; }
    }

    public string Color
    {
        get { return color; }
        set { color = value; }
    }

    public string Gender
    {
        get { return gender; }
        set { gender = value; }
    }

    public ClothingV3(string name, double price, string manufacturer, string id, string category, int stockQuantity, string size, string material, string color, string gender)
        : base(name, price, manufacturer, id, category, stockQuantity)
    {
        this.size = size;
        this.material = material;
        this.color = color;
        this.gender = gender;
    }

    public override void Display()
    {
        Console.WriteLine(GetInfo() + $", Размер: {Size}, Цвет: {Color}");
    }

    public string GetMaterialInfo()
    {
        return $"{Name}: материал — {Material}, для кого — {Gender}";
    }

    public void ChangeColor(string newColor)
    {
        Color = newColor;
        Console.WriteLine($"{Name}: цвет изменён на {Color}.");
    }

    public bool IsUnisex()
    {
        return Gender == "Унисекс";
    }

    public void Ship(string address)
    {
        Console.WriteLine($"{Name}: одежда отправлена по адресу {address}.");
    }

    public double CalculateShippingCost()
    {
        return 300;
    }
}

class BooksV3 : ProductV3, IShippable, IReviewable
{
    private string author;
    private string genre;
    private int pages;
    private int publicationYear;
    private List<string> reviews;

    public string Author
    {
        get { return author; }
        set { author = value; }
    }

    public string Genre
    {
        get { return genre; }
        set { genre = value; }
    }

    public int Pages
    {
        get { return pages; }
        set { pages = value < 0 ? 0 : value; }
    }

    public int PublicationYear
    {
        get { return publicationYear; }
        set { publicationYear = value; }
    }

    public BooksV3(string name, double price, string manufacturer, string id, string category, int stockQuantity, string author, string genre, int pages, int publicationYear)
        : base(name, price, manufacturer, id, category, stockQuantity)
    {
        this.author = author;
        this.genre = genre;
        this.pages = pages;
        this.publicationYear = publicationYear;
        this.reviews = new List<string>();
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Автор: {Author}, Жанр: {Genre}";
    }

    public string GetReadingTime()
    {
        double hours = Pages / 40.0;
        return $"{Name}: примерно {hours:F1} ч чтения.";
    }

    public bool IsModern()
    {
        return PublicationYear >= 2000;
    }

    public string GetAgeCategory()
    {
        if (PublicationYear < 1900) return "Классика";
        if (PublicationYear < 2000) return "Прошлый век";
        return "Современная";
    }

    public void Ship(string address)
    {
        Console.WriteLine($"{Name}: книга отправлена по адресу {address}.");
    }

    public double CalculateShippingCost()
    {
        return 200;
    }

    public void AddReview(string review)
    {
        reviews.Add(review);
        Console.WriteLine($"{Name}: отзыв добавлен.");
    }

    public List<string> GetReviews()
    {
        return reviews;
    }
}

class SmartphoneV3 : ElectronicsV3, IShippable, IReviewable, IWarrantyService
{
    private double screenSize;
    private int cameraMegapixels;
    private string os;
    private int batteryCapacity;
    private List<string> reviews;

    public double ScreenSize
    {
        get { return screenSize; }
        set { screenSize = value < 0 ? 0 : value; }
    }

    public int CameraMegapixels
    {
        get { return cameraMegapixels; }
        set { cameraMegapixels = value < 0 ? 0 : value; }
    }

    public string Os
    {
        get { return os; }
        set { os = value; }
    }

    public int BatteryCapacity
    {
        get { return batteryCapacity; }
        set { batteryCapacity = value < 0 ? 0 : value; }
    }

    public SmartphoneV3(string name, double price, string manufacturer, string id, string category, int stockQuantity, int warrantyPeriod, int powerConsumption, string countryOfOrigin, bool isSmart, double screenSize, int cameraMegapixels, string os, int batteryCapacity)
        : base(name, price, manufacturer, id, category, stockQuantity, warrantyPeriod, powerConsumption, countryOfOrigin, isSmart)
    {
        this.screenSize = screenSize;
        this.cameraMegapixels = cameraMegapixels;
        this.os = os;
        this.batteryCapacity = batteryCapacity;
        this.reviews = new List<string>();
    }

    public override string GetInfo()
    {
        return base.GetInfo() + $", Экран: {ScreenSize}\", Камера: {CameraMegapixels} Мп, ОС: {Os}";
    }

    public void TakePhoto()
    {
        Console.WriteLine($"{Name}: щёлк! Фото на {CameraMegapixels} Мп готово!!!");
    }

    public string CheckBattery()
    {
        return $"{Name}: батарея на {BatteryCapacity} мА·ч.";
    }

    public string GetSpecs()
    {
        return $"Экран {ScreenSize}\", {Os}, батарея {BatteryCapacity} мА·ч";
    }

    public void Ship(string address)
    {
        Console.WriteLine($"{Name}: смартфон отправлен по адресу {address}.");
    }

    public double CalculateShippingCost()
    {
        return 500;
    }

    public void AddReview(string review)
    {
        reviews.Add(review);
        Console.WriteLine($"{Name}: отзыв добавлен.");
    }

    public List<string> GetReviews()
    {
        return reviews;
    }

    public void RequestWarrantyRepair()
    {
        Console.WriteLine($"{Name}: заявка на гарантийный ремонт принята.");
    }

    public string GetWarrantyStatus()
    {
        return $"{Name}: гарантия {WarrantyPeriod} мес.";
    }
}

class StoreV3
{
    private string storeName;
    private string location;
    private List<ProductV3> products;

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }

    public string Location
    {
        get { return location; }
        set { location = value; }
    }

    public List<ProductV3> Products
    {
        get { return products; }
        set { products = value; }
    }

    public StoreV3(string storeName, string location)
    {
        this.storeName = storeName;
        this.location = location;
        products = new List<ProductV3>();
    }

    public void AddProduct(ProductV3 product)
    {
        products.Add(product);
    }

    public void ApplyDiscountToAll(double percent)
    {
        foreach (ProductV3 product in products)
        {
            product.Discount(percent);
        }
    }

    public void ShowAllProducts()
    {
        Console.WriteLine($"Магазин: {StoreName} ({Location})");
        foreach (ProductV3 product in products)
        {
            product.Display();
        }
    }

    public ProductV3 FindProductById(string id)
    {
        foreach (ProductV3 product in products)
        {
            if (product.Id == id) return product;
        }
        return null;
    }

    public double GetTotalInventoryValue()
    {
        double total = 0;
        foreach (ProductV3 product in products)
        {
            total += product.Price * product.StockQuantity;
        }
        return total;
    }

    public int GetProductCount()
    {
        return products.Count;
    }
}

SmartphoneV3 mySmartphoneV3 = new SmartphoneV3("Pixel 8", 70000, "Google", "SM-001", "Смартфон", 15, 24, 30, "Китай", true, 6.2, 50, "Android 14", 4575);
BooksV3 myBookV3 = new BooksV3("Война и мир", 1200, "ООО рога и копыта", "BK-001", "Книга", 8, "Лев Толстой", "Роман", 1225, 1869);
ClothingV3 myClothingV3 = new ClothingV3("Футболка", 1500, "Nike", "CL-001", "Одежда", 30, "L", "Хлопок", "Белый", "Унисекс");
ElectronicsV3 myElectronicsV3 = new ElectronicsV3("Телевизор", 50000, "SaSun", "EL-001", "Электроника", 5, 36, 150, "Корея", true);

StoreV3 myStoreV3 = new StoreV3("Товары V3", "ул. Ленина, 10");
myStoreV3.AddProduct(mySmartphoneV3);
myStoreV3.AddProduct(myBookV3);
myStoreV3.AddProduct(myClothingV3);
myStoreV3.AddProduct(myElectronicsV3);
myStoreV3.ShowAllProducts();

Console.WriteLine("--- Применяем скидку 10% ---");
myStoreV3.ApplyDiscountToAll(10);
myStoreV3.ShowAllProducts();

Console.WriteLine("--- Дополнительные возможности ---");
mySmartphoneV3.TakePhoto();
mySmartphoneV3.ToggleSmartMode();
Console.WriteLine(mySmartphoneV3.CheckBattery());
Console.WriteLine(myBookV3.GetReadingTime());
Console.WriteLine(myBookV3.GetAgeCategory());
myClothingV3.ChangeColor("Чёрный");
myClothingV3.Restock(20);
Console.WriteLine(mySmartphoneV3.GetStockStatus());

Console.WriteLine("--- Множественное наследование через интерфейсы ---");
IShippable[] shippableItems = { mySmartphoneV3, myBookV3, myClothingV3 };
foreach (IShippable item in shippableItems)
{
    item.Ship("г. Москва, ул. Пушкина, 10");
    Console.WriteLine($"Стоимость доставки: {item.CalculateShippingCost()} руб.");
}

IReviewable[] reviewableItems = { mySmartphoneV3, myBookV3 };
foreach (IReviewable item in reviewableItems)
{
    item.AddReview("Отличный товар!!!");
}

IWarrantyService[] warrantyItems = { mySmartphoneV3 };
foreach (IWarrantyService item in warrantyItems)
{
    item.RequestWarrantyRepair();
}

Console.WriteLine($"Всего товаров в магазине: {myStoreV3.GetProductCount()}");
Console.WriteLine($"Общая стоимость склада: {myStoreV3.GetTotalInventoryValue()} руб.");
ProductV3 foundProduct = myStoreV3.FindProductById("BK-001");
if (foundProduct != null)
{
    Console.WriteLine($"Найден товар: {foundProduct.Name}");
}
```
